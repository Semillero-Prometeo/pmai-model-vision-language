## La universidad

```python
from bs4 import BeautifulSoup
import requests

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3   ",
    "accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "accept-language": "en-US,en;q=0.5",
    "accept-encoding": "gzip, deflate, br",
    "connection": "keep-alive",
    "upgrade-insecure-requests": "1",   
}

url = "https://www.unilibre.edu.co/la-universidad"

respuesta = requests.get(url, headers=headers)

soup = BeautifulSoup(respuesta.content, "html.parser")

textos_principales = soup.find(id="brxe-ocglrx")
textos_principales_extraidos = textos_principales.find_all("div", class_="brxe-icon-box")

for contenedor in textos_principales_extraidos:
    parrafos = contenedor.find_all("p")
    for p in parrafos:
        print(p.text)
```

```python
from bs4 import BeautifulSoup
import requests

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3",
    "accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "accept-language": "en-US,en;q=0.5",
}

url = "https://www.unilibre.edu.co/la-universidad"
respuesta = requests.get(url, headers=headers)
soup = BeautifulSoup(respuesta.content, "html.parser")


elementos_brxe = soup.find_all(class_=lambda x: x and "brxe" in x)
print(f"Total elementos con clase 'brxe': {len(elementos_brxe)}")
print("\nClases únicas encontradas:")
clases_unicas = set()
for el in elementos_brxe:
    if el.get('class'):
        clases_unicas.add(" ".join(el.get('class')))
for clase in sorted(clases_unicas)[:20]:
    print(f"  - {clase}")

```

```python
import json

from bs4 import BeautifulSoup
import requests

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3",
    "accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8",
    "accept-language": "en-US,en;q=0.5",
}

url = "https://www.unilibre.edu.co/la-universidad"
respuesta = requests.get(url, headers=headers)
soup = BeautifulSoup(respuesta.content, "html.parser")


parrafos = soup.find_all("p")

for i, p in enumerate(parrafos, 1):
    texto = p.get_text(strip=True)
    print(f"\n{i}. {texto}") 


with open("data_unilibre.json", "w", encoding="utf-8") as f:
    json.dump(texto, f, ensure_ascii=False, indent=2)
```
